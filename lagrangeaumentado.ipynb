```python
import numpy as np
from scipy.optimize import minimize

# Definición de la función objetivo f(x1, x2)
def f(x):
    x1, x2 = x
    return (x1 - 1) ** 2 + (x2 - 2) ** 2

# Definición de la restricción h(x1, x2) = x1 + x2 - 3
def h(x):
    x1, x2 = x
    return x1 + x2 - 3

# Definición de la Lagrangiana aumentada L_A(x1, x2, λ, ρ)
def lagrange_augmented(x, lambd, rho):
    return f(x) + lambd * h(x) + (rho / 2) * (h(x) ** 2)

# Función para minimizar usando el Método de Lagrange Aumentado
def lagrange_method(x_init, lambd=0, rho=1, gamma=2, tol=1e-5, max_iter=10):
    x = np.array(x_init)
    for k in range(max_iter):
        # Minimización de la Lagrangiana aumentada
        result = minimize(lagrange_augmented, x, args=(lambd, rho), method='BFGS')
        x = result.x  # Nuevos valores de x1 y x2 obtenidos de la minimización

        # Mostrar progreso
        print(f"Iteración {k+1}:")
        print(f"Posición actual: x1 = {x[0]:.4f}, x2 = {x[1]:.4f}, Lagrange (λ) = {lambd:.4f}, Penalización (ρ) = {rho}")

        # Actualización del multiplicador de Lagrange y del parámetro ρ
        lambd = lambd - rho * h(x)
        rho *= gamma

        # Condición de parada cuando la restricción es aproximadamente 0
        if abs(h(x)) < tol:
            print("Restricción satisfecha y mínimo encontrado.")
            break
    
    return x

# Valores iniciales para x1, x2 y parámetros del método
x_init = [0, 0]
sol_x = lagrange_method(x_init)

print(f"Solución óptima: x1 = {sol_x[0]:.4f}, x2 = {sol_x[1]:.4f}")
```
